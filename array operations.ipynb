#### deeep copy and shallow copy

```python
# slicing
slice=a[:6]
slice=slice*2
slice
# there is no chnage in original array 
```

```python
b=a
b[0]=99
b
```

```python
a
# in case of slicing teh original does not chnage but in case of copying whole array and then changing second one then it will change 
```

# advance array manipulation

### stacking 

```python
a=np.array([1,2,3,4,5])
# we cannnot use matrices , here we have to use vectors 
```

```python
b=np.array([6,7,8,9,5])
```

```python
np.vstack((a,b))
# it will stack bvertically 
```

```python
np.hstack((a,b))
# horizontal stack
```

```python
np.column_stack((a,b))
```

### spliting 

```python
c=np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2)
```

```python
a=np.vsplit(c,4)
# it will split it in 4 equal parts
a
```

```python
for i in a:
    print(i)

# it will print the 4 arrays in a 
```

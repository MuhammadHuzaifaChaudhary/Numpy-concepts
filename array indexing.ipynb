```python
arr[0:2,2:4]
# in slicing we know the last index is not included 
```

```python
arr[:,2]
# slicing a column just like we did arr[0,:] for rows 
# : means that we want all items of column or row
# arr[:,2] it means we want all rows and 1st column ... because column contain all rows
```

```python
arr
```

```python
# doing jumping
arr[::2,::3]
# single colon means all elements but we dont want all elements we want jump of two elements like 
# oth row 2nd row and 4th row ... similary for columns we do this 
```

```python
arr[1,::3] # means first row but 3 jump elements
```

# slicing in 3d 

```python
arr3=np.arange(27).reshape(3,3,3)
# means that there are 3 2d arrays 
# first 3 is showing no of arrays middle is showing rows and last is shwoing columns 
arr3
```

```python
arr3[0,1,:]
# here 0 is showing first 2d array 
#  1 is shwoing the 2nd row odf first 2d array and colon is shwoing all columns 
```

```python
arr3[1,:,1]
# only 2nd column of 2nd matrix
```

### jumping in matrices in 3d tensor or for choosing 2 matrices the method is same 

```python
# suppose if we want to select first matrix and last matrix 
arr3[0::2,0:2,0:2]
# hence for jumping we have to use our previous concept using 2 colons 
# as we know the first element like 0::2 determines our matrices 
```
